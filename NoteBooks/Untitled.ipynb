```python
from sklearn.linear_model import LinearRegression
import pandas as pd
import numpy as np

df = pd.read_csv("/home/madhav/workspace/projects/yantragya/linear_regress.csv")

X = np.array(df["X"].dropna()).reshape(-1,1)
y = np.array(df["y"].fillna(0)).reshape(-1,1)


model = LinearRegression(fit_intercept=False)
model.fit(X, y)

print(model.coef_)
```
