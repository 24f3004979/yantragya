## Sampling

Sampling from a distribution is an important skill. While learning ML, it is often a good idea to work with toy-datasets. These toy-datasets can be generated from well known distributions such as the Bernoulli and Gaussian. The next section will introduce the idea of sampling in NumPy.

What happens if we increase the size of the dataset to $10000$? We see that `np.integers` is only giving us observations from a discrete uniform distribution. In other words, if you think about it as a coin toss, the coin is unbiased. How do we generate samples from a biased coin or a Bernoulli distribution with $p \neq 0.5$? We use a method called `choice`. This method "chooses" numbers from a list or an array based on the probabilities given to it in the array `p`.

```python
rng = np.random.default_rng(seed = 1001)
X = rng.choice([0, 1], size = 1000, p = [0.2, 0.8])
# Specific choice for the given number generation

# I would be predicting the given probability number for the given condition for one
p = np.sum(X) / len(X)
print(f'P value being computed with probability to match to 0.8 : {p} | Holds true')
```

Note that the array `p` specifies the probabilities of $0$ and $1$.

## Problem-7

Consider the following Gaussian distribution:

$$
\mathcal{N}(3, 1.5^2)
$$

Compute the density of the point $5$ with respect to this distribution. Enter your answer correct to three decimal places.

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# 1. Generate Gaussian data
# Parameters: mean (mu), standard deviation (sigma), number of samples
mu = 3       
sigma = 1.5   
samples = 100000

data = np.random.normal(loc=mu, scale=sigma, size=samples)
df = pd.DataFrame(data)

print(df.head())

# 2. Plot the data as a histogram
# 'density=True' normalizes the histogram so the total area equals 1
plt.hist(data, bins=100, density=True, alpha=0.6, color='g', edgecolor='black')

# 3. Add labels and title
plt.title('Gaussian Distribution (Mean=0, SD=1)')
plt.xlabel('Value')
plt.ylabel('Probability Density')
plt.grid(True, alpha=0.3)

plt.show()
```

## Problem-10

Plot the beta distribution for the following values:

- $\alpha = 5, \beta = 5$
- $\alpha = 5, \beta = 1.5$
- $\alpha = 1.5, \beta = 5$
- $\alpha = 0.5, \beta = 0.5$


Do not worry about the normalization constant. Use `plt.subplot(2, 2, i)` to plot all four distributions in a single figure. `i` should take the values 1, 2, 3 and 4 for these four subplots. If you are not sure how it works, google it. Stackoverflow will be of great help here.

```python
import numpy as np 
import matplotlib.pyplot as plt

# Keeping global sample size
sample_size = 1000000
bin_number = 50

x1 = np.random.beta(5, 5, sample_size)
x2 = np.random.beta(5, 15, sample_size)
x3 = np.random.beta(1.5, 5, sample_size)
x4 = np.random.beta(0.5, 0.5, sample_size)

# Visualization for the given distribution
colors = ['red', 'blue', 'green', 'purple']
labels = [ 'centered', 'left_skqwed', 'extream_left sqwed', 'spreaded to both sides']

i = 0
lst = [x1,x2,x3,x4]
data_sets = []
while i < 4:
    elem = {
        "data": lst[i],
        "color": colors[i],
        "label": labels[i]
    }
    data_sets.append(elem)
    i += 1
print(f"Data set for loading for visualization : {data_sets}")

# plot loop
plt.figure(figsize=(10,6))

# Iteration for making plot
for d in data_sets:
    plt.hist(
        d['data'],
        bins=bin_number,
        density=True,
        alpha=0.5,
        edgecolor=None,
        color=d['color'],
        label=d['label']
    )
plt.title("Beta Distribution Showdown")
plt.legend(fontsize=11)
plt.show()
```
