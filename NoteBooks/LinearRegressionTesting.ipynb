```python
#@title Random Numbers

import matplotlib.pyplot as plt
import numpy as np

np.random.seed(40)

# random number with target output
slope = 3
total_points = 50
intercept = 2

X = np.linspace(0,10, total_points)

y_linear = (slope * X) + intercept

noise = np.random.normal(loc=0, scale=2.0, size=len(X))
Y = (slope * X) + intercept + noise

# Loaded dataset
print(X, Y)
```

```python
#@title : Making Equation and solving for Weights
stacking = np.ones(len(X))

# Initial A matrix for the final computation
A = np.column_stack((X,stacking))

transpose_product = np.matmul(A.T, A)
inv = np.linalg.inv(transpose_product)

# x.t y
label_x = np.matmul(A.T, Y)

final_weights = np.dot(inv, label_x)

print(final_weights)  # works with butifuly
```

```python
#@title Final render
import matplotlib.pyplot as plt
# Making final render for the Linear Regression model

plt.scatter(X,Y)

# Making Line render with final_predicted weights
slope = final_weights[0]
intercept = final_weights[1]

# Ax line Draw
plt.axline(xy1=(0,intercept), slope=slope, color='red')

plt.plot()
plt.xlabel('Input set')
plt.ylabel('Target')
plt.title('Linear Regression Model')
```
